```python
def process_video_with_stats(video_path, landmarks_seq, output_path):
    cap = cv2.VideoCapture(video_path)
    fps = cap.get(cv2.CAP_PROP_FPS)
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))

    punches_made, punches_missed = 0, 0
    blocks_made, blocks_missed = 0, 0
    score = 0.0
    last_punch_result = "made"

    punch_flags = detect_punch(landmarks_seq)

    features_seq = extract_pose_features(landmarks_seq)

    idx = 0
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        # Update stats
        if punch_flags[idx]:
            punches_made += 1
            score += 1.0
            last_punch_result = "made"
        else:
            punches_missed += 1
            last_punch_result = "missed"

        # Overlay stats
        f = features_seq[idx]
        frame = overlay_stats(
            frame,
            score=score,
            punches_made=punches_made,
            punches_missed=punches_missed,
            blocks_made=blocks_made,
            blocks_missed=blocks_missed,
            kinematic=str(f["kinematic"]),
            planar=str(f["planar"]),
            volumetric=str(f["volumetric"]),
            overall="OK",
            last_punch_result=last_punch_result,
            timestamp=idx/fps
        )

        out.write(frame)
        idx += 1

    cap.release()
    out.release()
```

```python
import torch
import torch.nn as nn

class CNNPunchDetector(nn.Module):
    """
    Deeper MLP over flattened landmarks with batchnorm + dropout.
    """
    def __init__(self, input_dim, hidden1=256, hidden2=128, dropout=0.3):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden1),
            nn.BatchNorm1d(hidden1),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden1, hidden2),
            nn.BatchNorm1d(hidden2),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden2, 2)
        )
    def forward(self, x): return self.net(x)

class PunchBiLSTM(nn.Module):
    """
    Bidirectional LSTM with dropout and last-timestep pooling.
    """
    def __init__(self, input_dim, hidden_dim=128, num_layers=2, bidirectional=True, dropout=0.3, num_classes=2):
        super().__init__()
        self.lstm = nn.LSTM(
            input_dim, hidden_dim, num_layers,
            batch_first=True, dropout=dropout, bidirectional=bidirectional
        )
        out_dim = hidden_dim * (2 if bidirectional else 1)
        self.fc = nn.Linear(out_dim, num_classes)
    def forward(self, x):
        out, _ = self.lstm(x)        # (B, T, H*)
        out = out[:, -1, :]          # last timestep
        return self.fc(out)
```

```python
def relabel_sequences_any_punch(labels, seq_len=15):
    """
    Label a sequence positive if it contains ANY punch frame (instead of >=2).
    """
    y_seq = []
    for i in range(len(labels) - seq_len):
        y_seq.append(1 if sum(labels[i:i+seq_len]) >= 1 else 0)
    return np.array(y_seq, dtype=np.int64)

def build_sequences(landmarks, seq_len=15):
    X_seq, centers = [], []
    for i in range(len(landmarks) - seq_len):
        X_seq.append(np.stack([landmarks[j].flatten() for j in range(i, i+seq_len)], axis=0))
        centers.append(i + seq_len // 2)
    return np.array(X_seq, dtype=np.float32), centers
```

```python
def detect_punch_wrist_forward(landmarks_seq, vel_thr=0.05, forward_margin=0.01):
    flags = []
    for i, lm in enumerate(landmarks_seq):
        if lm is None:
            flags.append(False); continue

        torso_center_x = np.mean([lm[11,0], lm[12,0], lm[23,0], lm[24,0]])
        lwx, rwx = lm[15,0], lm[16,0]

        left_forward = lwx < (torso_center_x - forward_margin)
        right_forward = rwx < (torso_center_x - forward_margin)

        if i > 0 and landmarks_seq[i-1] is not None:
            prev = landmarks_seq[i-1]
            v_left = np.linalg.norm(lm[15,:3] - prev[15,:3])
            v_right = np.linalg.norm(lm[16,:3] - prev[16,:3])
        else:
            v_left = v_right = 0.0

        punch_detected = ((left_forward and v_left > vel_thr) or
                          (right_forward and v_right > vel_thr))

        # Add cooldown logic: only count punch if previous frame was not a punch
        if i > 0 and flags[-1] and punch_detected:
            punch_detected = False

        flags.append(punch_detected)
    return flags
```

```python
def overlay_stats(
    display_frame, score, punches_made, punches_missed,
    blocks_made, blocks_missed,
    kinematic, planar, volumetric, overall,
    punch_metrics=None,
    last_punch_result="made",
    current_color=(0,255,0),
    subtitle=None,
    timestamp=None
):
    stats_font = cv2.FONT_HERSHEY_SIMPLEX
    stats_border = (0,0,0); stats_scale=0.8
    stats_thickness=2; stats_border_thickness=4
    stats_spacing=22; white_color=(255,255,255)
    h,w = display_frame.shape[:2]
    stats_x=10; stats_y=int(h*0.05)

    # Score
    cv2.putText(display_frame, f"Score: {score:.2f}", (stats_x, stats_y),
                stats_font, stats_scale, stats_border, stats_border_thickness, cv2.LINE_AA)
    cv2.putText(display_frame, f"Score: {score:.2f}", (stats_x, stats_y),
                stats_font, stats_scale, white_color, stats_thickness, cv2.LINE_AA)

    # Punch counts
    made_text = f"Punches Made: {punches_made}"
    cv2.putText(display_frame, made_text, (stats_x, stats_y+stats_spacing),
                stats_font, stats_scale, stats_border, stats_border_thickness, cv2.LINE_AA)
    cv2.putText(display_frame, made_text, (stats_x, stats_y+stats_spacing),
                stats_font, stats_scale,
                current_color if last_punch_result=="made" else white_color,
                stats_thickness, cv2.LINE_AA)

    missed_text = f"Punches Missed: {punches_missed}"
    cv2.putText(display_frame, missed_text, (stats_x, stats_y+2*stats_spacing),
                stats_font, stats_scale, stats_border, stats_border_thickness, cv2.LINE_AA)
    cv2.putText(display_frame, missed_text, (stats_x, stats_y+2*stats_spacing),
                stats_font, stats_scale,
                current_color if last_punch_result=="missed" else white_color,
                stats_thickness, cv2.LINE_AA)

    # Biomechanics
    extra_stats = [
        f"Kinematic: {kinematic}",
        f"Planar: {planar}",
        f"Volumetric: {volumetric}",
        f"Overall: {overall}",
    ]
    for idx,text in enumerate(extra_stats,start=3):
        cv2.putText(display_frame, text, (stats_x, stats_y+idx*stats_spacing),
                    stats_font, stats_scale, stats_border, stats_border_thickness, cv2.LINE_AA)
        cv2.putText(display_frame, text, (stats_x, stats_y+idx*stats_spacing),
                    stats_font, stats_scale, white_color, stats_thickness, cv2.LINE_AA)

    # Punch-specific metrics
    if punch_metrics is not None:
        punch_texts = [
            f"LW Dist: {punch_metrics['left_wrist_dist']:.2f}",
            f"RW Dist: {punch_metrics['right_wrist_dist']:.2f}",
            f"LW Vel: {punch_metrics['left_wrist_vel']:.2f}",
            f"RW Vel: {punch_metrics['right_wrist_vel']:.2f}",
        ]
        for idx,text in enumerate(punch_texts,start=7):
            cv2.putText(display_frame, text, (stats_x, stats_y+idx*stats_spacing),
                        stats_font, stats_scale, stats_border, stats_border_thickness, cv2.LINE_AA)
            cv2.putText(display_frame, text, (stats_x, stats_y+idx*stats_spacing),
                        stats_font, stats_scale, (0,200,255), stats_thickness, cv2.LINE_AA)

    # Timestamp and subtitle
    if timestamp is not None:
        cv2.putText(display_frame, f"Time: {timestamp:.2f}s",
                    (stats_x, stats_y+11*stats_spacing),
                    stats_font, stats_scale, white_color, stats_thickness, cv2.LINE_AA)
    if subtitle is not None:
        cv2.putText(display_frame, subtitle,
                    (stats_x, stats_y+12*stats_spacing),
                    stats_font, stats_scale, (200,200,0), stats_thickness, cv2.LINE_AA)

    return display_frame
```

## Cross-Validation & Held-Out Test Set Evaluation

This cell implements best practices for model assessment using proper train/val/test splits and k-fold cross-validation to ensure robust evaluation.

```python
from sklearn.model_selection import KFold, StratifiedKFold
from sklearn.metrics import precision_recall_fscore_support, precision_score, recall_score, accuracy_score, classification_report

print("="*70)
print("CROSS-VALIDATION & HELD-OUT TEST SET EVALUATION")
print("="*70)

# ===== SETUP: Create Train/Val/Test Split =====
print("\n1. Creating Train/Val/Test Splits")
print("-" * 70)

# Use all sequence data
total_sequences = len(X_seq)
print(f"Total sequences available: {total_sequences}")
print(f"Positive sequences: {sum(y_seq)} ({np.mean(y_seq)*100:.1f}%)")
print(f"Negative sequences: {len(y_seq) - sum(y_seq)} ({(1-np.mean(y_seq))*100:.1f}%)")

# Create train(60%) / val(20%) / test(20%) split with stratification
train_split = int(0.6 * len(X_seq))
temp_split = int(0.8 * len(X_seq))

X_train_fold = X_seq[:train_split]
y_train_fold = y_seq[:train_split]
X_val_fold = X_seq[train_split:temp_split]
y_val_fold = y_seq[train_split:temp_split]
X_test_fold = X_seq[temp_split:]
y_test_fold = y_seq[temp_split:]

print(f"\nTrain set: {len(X_train_fold)} sequences ({np.mean(y_train_fold)*100:.1f}% positive)")
print(f"Val set:   {len(X_val_fold)} sequences ({np.mean(y_val_fold)*100:.1f}% positive)")
print(f"Test set:  {len(X_test_fold)} sequences ({np.mean(y_test_fold)*100:.1f}% positive)")

# ===== K-FOLD CROSS-VALIDATION (on training data) =====
print("\n2. K-Fold Cross-Validation on Training Data")
print("-" * 70)

n_splits = 5
skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)

fold_results = []
fold_num = 0

for fold_idx, (train_indices, val_indices) in enumerate(skf.split(X_train_fold, y_train_fold), 1):
    print(f"\nFold {fold_idx}/{n_splits}:")
    
    # Split training data for this fold
    X_fold_train = X_train_fold[train_indices]
    y_fold_train = y_train_fold[train_indices]
    X_fold_val = X_train_fold[val_indices]
    y_fold_val = y_train_fold[val_indices]
    
    print(f"  Train: {len(X_fold_train)} ({np.mean(y_fold_train)*100:.1f}% pos) | Val: {len(X_fold_val)} ({np.mean(y_fold_val)*100:.1f}% pos)")
    
    # Create datasets and loaders for this fold
    fold_train_ds = SequenceDataset(X_fold_train, y_fold_train, augment=True, noise_std=0.01)
    fold_val_ds = SequenceDataset(X_fold_val, y_fold_val, augment=False)
    
    fold_sampler = make_weighted_sampler(y_fold_train)
    fold_train_loader = DataLoader(fold_train_ds, batch_size=32, sampler=fold_sampler)
    fold_val_loader = DataLoader(fold_val_ds, batch_size=32, shuffle=False)
    
    # Train LSTM for this fold
    device = "cuda" if torch.cuda.is_available() else "cpu"
    fold_lstm = PunchBiLSTM(input_dim=student_landmarks_seq[0].size, hidden_dim=128, num_layers=2, bidirectional=True, dropout=0.3)
    fold_lstm = train_with_early_stopping(fold_lstm, fold_train_loader, fold_val_loader, 
                                          epochs=30, lr=5e-4, class_weights=(1.0, 15.0), 
                                          device=device)
    
    # Evaluate on fold validation set
    fold_eval = evaluate_classifier(fold_lstm, fold_val_loader, device=device, threshold=0.4, save_prefix=f"fold{fold_idx}_lstm")
    f1_fold = f1_score(fold_eval['y_true'], fold_eval['y_pred'], average='binary')
    precision_fold = precision_score(fold_eval['y_true'], fold_eval['y_pred'], average='binary', zero_division=0)
    recall_fold = recall_score(fold_eval['y_true'], fold_eval['y_pred'], average='binary', zero_division=0)
    
    fold_results.append({
        'fold': fold_idx,
        'f1': f1_fold,
        'precision': precision_fold,
        'recall': recall_fold,
        'y_true': fold_eval['y_true'],
        'y_pred': fold_eval['y_pred']
    })
    
    print(f"  Fold F1: {f1_fold:.3f} | Precision: {precision_fold:.3f} | Recall: {recall_fold:.3f}")

# Print cross-validation summary
print(f"\n{'='*70}")
print(f"CROSS-VALIDATION SUMMARY (K={n_splits})")
print(f"{'='*70}")
f1_scores = [r['f1'] for r in fold_results]
f1_strs = [f'{x:.3f}' for x in f1_scores]
prec_strs = [f'{r["precision"]:.3f}' for r in fold_results]
recall_strs = [f'{r["recall"]:.3f}' for r in fold_results]
print(f"F1 Scores:   {f1_strs}")
print(f"Mean F1:     {np.mean(f1_scores):.3f} (±{np.std(f1_scores):.3f})")
print(f"Precision:   {prec_strs}")
print(f"Mean Prec:   {np.mean([r['precision'] for r in fold_results]):.3f} (±{np.std([r['precision'] for r in fold_results]):.3f})")
print(f"Recall:      {recall_strs}")
print(f"Mean Recall: {np.mean([r['recall'] for r in fold_results]):.3f} (±{np.std([r['recall'] for r in fold_results]):.3f})")

# ===== HELD-OUT TEST SET EVALUATION =====
print(f"\n{'='*70}")
print("HELD-OUT TEST SET EVALUATION (Best Model)")
print(f"{'='*70}")

# Train final model on full train+val set for test evaluation
X_train_final = np.concatenate([X_train_fold, X_val_fold], axis=0)
y_train_final = np.concatenate([y_train_fold, y_val_fold], axis=0)

train_final_ds = SequenceDataset(X_train_final, y_train_final, augment=True, noise_std=0.01)
final_sampler = make_weighted_sampler(y_train_final)
train_final_loader = DataLoader(train_final_ds, batch_size=32, sampler=final_sampler)

# Create dummy val loader for training (test set will be evaluated separately)
val_final_ds = SequenceDataset(X_test_fold, y_test_fold, augment=False)
val_final_loader = DataLoader(val_final_ds, batch_size=32, shuffle=False)

print(f"\nTraining on combined train+val set ({len(X_train_final)} sequences)")
print(f"Test set: {len(X_test_fold)} sequences ({np.mean(y_test_fold)*100:.1f}% positive)")

device = "cuda" if torch.cuda.is_available() else "cpu"
final_lstm = PunchBiLSTM(input_dim=student_landmarks_seq[0].size, hidden_dim=128, num_layers=2, bidirectional=True, dropout=0.3)
final_lstm = train_with_early_stopping(final_lstm, train_final_loader, val_final_loader,
                                       epochs=30, lr=5e-4, class_weights=(1.0, 15.0),
                                       device=device)

# Evaluate on held-out test set
test_loader = DataLoader(val_final_ds, batch_size=32, shuffle=False)
test_eval = evaluate_classifier(final_lstm, test_loader, device=device, threshold=0.4, save_prefix="lstm_test")

test_f1 = f1_score(test_eval['y_true'], test_eval['y_pred'], average='binary')
test_precision = precision_score(test_eval['y_true'], test_eval['y_pred'], average='binary', zero_division=0)
test_recall = recall_score(test_eval['y_true'], test_eval['y_pred'], average='binary', zero_division=0)
test_accuracy = accuracy_score(test_eval['y_true'], test_eval['y_pred'])

print(f"\n{'='*70}")
print("TEST SET RESULTS")
print(f"{'='*70}")
print(f"Accuracy:  {test_accuracy:.3f}")
print(f"Precision: {test_precision:.3f}")
print(f"Recall:    {test_recall:.3f}")
print(f"F1 Score:  {test_f1:.3f}")

# Detailed classification report
print(f"\nDetailed Classification Report (Test Set):")
print(classification_report(test_eval['y_true'], test_eval['y_pred'], 
                           labels=[0, 1], target_names=['No Punch', 'Punch']))

# ===== COMPARISON: CV vs Test Performance =====
print(f"\n{'='*70}")
print("CROSS-VALIDATION vs TEST PERFORMANCE")
print(f"{'='*70}")
cv_mean_f1 = np.mean(f1_scores)
cv_std_f1 = np.std(f1_scores)
print(f"CV Mean F1 (±std):   {cv_mean_f1:.3f} (±{cv_std_f1:.3f})")
print(f"Test F1:             {test_f1:.3f}")
print(f"Difference:          {abs(test_f1 - cv_mean_f1):.3f}")

if abs(test_f1 - cv_mean_f1) < cv_std_f1:
    print("✓ Test performance within expected range (no significant overfitting detected)")
else:
    print("⚠ Test performance differs from CV (possible overfitting or distribution shift)")

# ===== VISUALIZATION: Cross-validation Performance =====
print(f"\nGenerating cross-validation visualization...")

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# F1 Scores
fold_ids = [r['fold'] for r in fold_results]
f1_vals = [r['f1'] for r in fold_results]
axes[0].bar(fold_ids, f1_vals, color='steelblue', alpha=0.7)
axes[0].axhline(y=test_f1, color='red', linestyle='--', label=f'Test F1: {test_f1:.3f}', linewidth=2)
axes[0].set_xlabel('Fold')
axes[0].set_ylabel('F1 Score')
axes[0].set_title(f'F1 Across Folds (Mean: {cv_mean_f1:.3f}±{cv_std_f1:.3f})')
axes[0].set_ylim([0, 1])
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Precision
prec_vals = [r['precision'] for r in fold_results]
axes[1].bar(fold_ids, prec_vals, color='green', alpha=0.7)
axes[1].set_xlabel('Fold')
axes[1].set_ylabel('Precision')
axes[1].set_title(f'Precision Across Folds (Mean: {np.mean(prec_vals):.3f})')
axes[1].set_ylim([0, 1])
axes[1].grid(True, alpha=0.3)

# Recall
recall_vals = [r['recall'] for r in fold_results]
axes[2].bar(fold_ids, recall_vals, color='orange', alpha=0.7)
axes[2].set_xlabel('Fold')
axes[2].set_ylabel('Recall')
axes[2].set_title(f'Recall Across Folds (Mean: {np.mean(recall_vals):.3f})')
axes[2].set_ylim([0, 1])
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('cv_performance_summary.png', dpi=100, bbox_inches='tight')
plt.close()

print(f"✓ Visualization saved: cv_performance_summary.png")

print(f"\n{'='*70}")
print(" CROSS-VALIDATION & TEST EVALUATION COMPLETE")
print(f"{'='*70}")
```
